```
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
import glob
import numpy as np
from razdel import tokenize
import random
from collections import Counter
import matplotlib.pyplot as plt  # Для графиков

# -------------------- 1. Гиперпараметры --------------------
BATCH_SIZE = 32
MAX_WORDS = 10000
EPOCHS = 10
EMBEDDING_DIM = 300
DROPOUT_RATE = 0.5
LSTM_UNITS = 128
LEARNING_RATE = 1e-4
WIN_SIZE = 1000
WIN_STEP = 200
CLASS_LIST = ["Dostoevsky", "Tolstoy", "Turgenev", "Chekhov", "Lermontov"]
TEST_SPLIT = 0.2

# -------------------- 2. Функции --------------------

def load_data(class_list, prose_dir="./dataset/prose", poems_dir="./dataset/poems"):
    """Загружает тексты из файлов."""
    all_texts = {}
    for author in class_list:
        all_texts[author] = ''
        for path in glob.glob(f'{prose_dir}/{author}/*.txt') + glob.glob(f'{poems_dir}/{author}/*.txt'):
            with open(f'{path}', 'r', errors='ignore') as f:
                text = f.read()
            all_texts[author] += ' ' + text.replace('\n', ' ')
    return all_texts

def razdel_tokenize(text):
    """Токенизирует текст с использованием razdel."""
    return [token.text for token in tokenize(text)]

def create_vocabulary(all_texts, max_words):
    """Создает словарь на основе токенизированных текстов."""
    word_counts = {}
    for author, text in all_texts.items():
        tokens = razdel_tokenize(text)
        for token in tokens:
            if token not in word_counts:
                word_counts[token] = 0
            word_counts[token] += 1

    sorted_words = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)
    vocabulary = [word for word, count in sorted_words]
    vocabulary = vocabulary[:max_words-1]
    vocabulary.append("<unk>")
    word_to_index = {word: i for i, word in enumerate(vocabulary)}
    index_to_word = {i: word for word, i in word_to_index.items()}
    return word_to_index, index_to_word

def texts_to_indices(all_texts, word_to_index):
  """Преобразует тексты в последовательности индексов."""
  seq_all = {}
  for author, text in all_texts.items():
      tokens = razdel_tokenize(text)
      seq_all[author] = [word_to_index.get(token, word_to_index["<unk>"]) for token in tokens]
  return seq_all

def prepare_data(seq_all, class_list, test_split, sequence_length, random_state=42):
    """Разделяет данные на обучающую, валидационную и тестовую выборки."""
    x = []
    y = []
    for i, author in enumerate(class_list):
        seq = seq_all[author]
        x += seq
        y += [i] * len(seq)

    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=test_split, stratify=y, random_state=random_state)
    x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, test_size=0.125, stratify=y_train, random_state=random_state)

    print("Распределение классов в обучающей выборке:", Counter(y_train))
    print("Распределение классов в валидационной выборке:", Counter(y_val))
    print("Распределение классов в тестовой выборке:", Counter(y_test))

    return x_train, x_val, x_test, y_train, y_val, y_test

def create_windowed_dataset(x, y, win_size, step, num_classes):
    """Создает датасет с окнами."""
    x_windows = []
    y_categorical = []
    for i in range(0, len(x) - win_size + 1, step):
        x_windows.append(x[i:i + win_size])
        window_labels = y[i:i+win_size]
        most_common_label = max(set(window_labels), key=window_labels.count)
        y_categorical.append(keras.utils.to_categorical(most_common_label, num_classes=num_classes))
    return np.array(x_windows), np.array(y_categorical)

def load_embeddings(word_to_index, max_words, embedding_dim, embedding_file='navec_hudlit_v1_12B_500K_300d_100q.tar'):
    """Загружает предобученные эмбеддинги Navec."""
    from navec import Navec
    navec = Navec.load(embedding_file)
    embedding_matrix = np.zeros((max_words, embedding_dim))
    unk_vector = navec['<unk>'] if '<unk>' in navec else np.zeros(embedding_dim)

    for word, i in word_to_index.items():
        if i < max_words:
            try:
                embedding_vector = navec[word]
            except KeyError:
                embedding_vector = unk_vector
    embedding_matrix[i] = embedding_vector

    return embedding_matrix

def create_model(max_words, embedding_dim, embedding_matrix, dropout_rate, lstm_units, num_classes, win_size, learning_rate):
    """Создает модель LSTM."""
    from keras.models import Sequential
    from keras.layers import Embedding, SpatialDropout1D, BatchNormalization, Dropout, LSTM, Dense
    from keras.optimizers import Adam

    model = Sequential()
    model.add(Embedding(max_words, embedding_dim, input_length=win_size, weights=[embedding_matrix], trainable=False))
    model.add(SpatialDropout1D(dropout_rate))
    model.add(LSTM(lstm_units))
    model.add(BatchNormalization())
    model.add(Dropout(dropout_rate))
    model.add(Dense(num_classes, activation='softmax'))

    optimizer = Adam(learning_rate=learning_rate)
    model.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def plot_history(history):
    """Строит графики точности и потерь во время обучения."""
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'])
    plt.plot(history.history['val_accuracy'])
    plt.title('Точность модели')
    plt.xlabel('Эпоха')
    plt.ylabel('Точность')
    plt.legend(['Обучение', 'Валидация'], loc='upper left')

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'])
    plt.plot(history.history['val_loss'])
    plt.title('Потери модели')
    plt.xlabel('Эпоха')
    plt.ylabel('Потери')
    plt.legend(['Обучение', 'Валидация'], loc='upper right')

    plt.show()

# -------------------- 3. Подготовка данных --------------------
all_texts = load_data(CLASS_LIST)
word_to_index, index_to_word = create_vocabulary(all_texts, MAX_WORDS)
seq_all = texts_to_indices(all_texts, word_to_index)

# Балансировка данных
sequence_length = 40000
seq_balance = {}
for author in CLASS_LIST:
    seq_balance[author] = seq_all[author][:sequence_length]

x_train, x_val, x_test, y_train, y_val, y_test = prepare_data(seq_balance, CLASS_LIST, TEST_SPLIT, sequence_length)

x_train_win, y_train_cat = create_windowed_dataset(x_train, y_train, WIN_SIZE, WIN_STEP, len(CLASS_LIST))
x_val_win, y_val_cat = create_windowed_dataset(x_val, y_val, WIN_SIZE, WIN_STEP, len(CLASS_LIST))
x_test_win, y_test_cat = create_windowed_dataset(x_test, y_test, WIN_SIZE, WIN_STEP, len(CLASS_LIST))

# -------------------- 4. Загрузка эмбеддингов --------------------
embedding_matrix = load_embeddings(word_to_index, MAX_WORDS, EMBEDDING_DIM)

# -------------------- 5. Создание и обучение модели --------------------
model = create_model(MAX_WORDS, EMBEDDING_DIM, embedding_matrix, DROPOUT_RATE, LSTM_UNITS, len(CLASS_LIST), WIN_SIZE, LEARNING_RATE)

checkpoint = keras.callbacks.ModelCheckpoint('best_model.keras', monitor='val_accuracy', save_best_only=True, mode='max')
early_stopping = keras.callbacks.EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

history = model.fit(x_train_win, y_train_cat,
                    epochs=EPOCHS,
                    batch_size=BATCH_SIZE,
                    validation_data=(x_val_win, y_val_cat),
                    callbacks=[checkpoint, early_stopping])

# -------------------- 6. Оценка модели --------------------
loss, accuracy = model.evaluate(x_test_win, y_test_cat)
print('Test accuracy:', accuracy)

# -------------------- 7. Графики --------------------
plot_history(history)

# -------------------- 8. Сохранение модели --------------------
model.save("my_model.keras")
print("Модель сохранена в файл my_model.keras")
```
