Запрещается пользоваться библиотекой numpy. На вход подается список. Матричное умножение реализуется с помощью циклов самостоятельно. Класс содержит методы для вычисления взвешенной суммы, а также содержит функции активации из урока.

```python
class Neuron:
    def __init__ (self, w):
        self.w = w

    def onestep(self, x): # Функция Хевисайда
        if x >= 0:
            return 1
        else:
            return 0

    def у(self, x):
        s = 0
        for i in range(0,len(self.w)): # Сумматор
            s += self.w[i]*x[i]
        return s

bias = 10

# Задаем входной вектор
Xi = [1, 0, 0, 1, 1, 1]

# Задаем вектор весов
Wi = [5, 4, 1, 1, 6, 7]
n = Neuron(Wi)
S = n.у(Xi) - bias
print('S = ', n.у(Xi))
print('Y =', n.onestep(S))
```

Это типичная задача, в которой нужно проанализировать сочетание множества факторов и на основе этого анализа принять итоговое решение. Для простоты учтем всего шесть факторов, влияющих на покупку игрового монитора или же воздержаться на данный момент от его покупки. На входы в нейрон мы подадим следующие исходные данные и оценим влияние этих факторов решение о покупке монитора:

1.   x1 — наличие свободных денег на покупку;
2.   x2 — наличие определенного монитора в магазине;
3.   x3 — мощность компьютера;
4.   x4 — отзывы о магазине;
5.   x5 — отзывы об игровом мониторе;
6.   x6 — обстановка на рынке техники.

# Описание исходных данных и почему я выбрал именно их:
1. Наличие свободных денег на покупку (x1):
Этот параметр является ключевым. Если нет достаточного бюджета, то мы не сможем купить желаемый нами монитор, даже если все остальные условия благоприятны. Необходимо заранее определить, сколько денег нужно потратить и готовы ли мы к таким тратам.

2. Наличие определенного монитора в магазине (x2):
Даже если вы нашли подходящий монитор и у вас есть деньги, он может не быть в наличии в магазине. Это может привести к необходимости поиска альтернативных моделей или покупки в другом месте, что может занять время.

3. Мощность компьютера (x3):
Мощность вашего компьютера будет определять, насколько хорошо он сможет «развернуть» возможности игрового монитора. Например, если у вас есть монитор с высоким разрешением и частотой обновления, но ваш компьютер не справляется с их обработкой, то и покупать игровой монитор довольно бессмысленно.
4. Отзывы о магазине (x4):
Отзывы о магазине важны для выбора надежного продавца. Если магазин имеет плохие отзывы, это может вызвать сомнения в качестве обслуживания, наличия гарантийного обслуживания и возможность возврата товара. Качественный сервис и наличие надежной репутации могут стать решающими факторами при выборе места покупки.
5. Отзывы об игровом мониторе (x5):
Отзывы о конкретном мониторе помогут вам понять его качества: яркость, цветопередачу, время отклика и наличие проблем с дизайном или функционалом. Положительные отзывы могут закрепить ваше желание купить именно эту модель, в то время как отрицательные могут оттолкнуть от покупки. Они помогут определиться, так ли хороша данная модель монитора или нет.

6. Ситуация на рынке техники (x6):
Рынок технологий постоянно меняется: бывают акции, распродажи, выход новых моделей и изменения в ценах. Если на рынке наблюдается снижение цен или вы ожидаете выход новой модели, это может повлиять на ваше решение о покупке сейчас или ожидании более выгодного предложения.

Для упрощения нашей модели присвоим всем этим параметрам логические значения в виде цифр 0 или 1. Например, если есть деньги на покупку игрового монитора и это ни как не отразится на финансовое состояние в целом, то можно покупать значение будет 1, если же ситуация обратная, то от покупки стоит воздержаться и значение уже будет 0, соответственно и далее, если есть определенный монитор в магазине то 1, если нет, то 0, является ли мощным ваш компьютер, если да то 1, иначе 0, хорошие ли отзывы в магазине, в которым вы хотите приобрести монитор, если да то 1, нет то 0, аналогично и с отзывами о мониторе, хорошие и объективные, то 1, иначе 0, одной из важной составляющей является ситуация на рынке техники, если она благоприятная, то 1, если у вас вызывает сомнение и вы решили подождать более лучшие модели или выгодное предложение, то 0.   

Если у нейрона есть шесть входов, то должны быть и шесть весовых коэффициентова. В нашем примере весовые коэффициенты можно представить как показатели важности каждого входа, влияющие на общее решение нейрона. Чем больше значение коэффициента, тем больше важность входного параметра. Веса входов распределим следующим образом:

1.   w1=5

Мы задали весовые коэффициенты таким образом, что важную роль играли факторы 1, 3, 6 (наличие свободных денег на покупку, мощность компьютера, обстановка на рынке техники). Они же и будут самыми важными при принятии нейроном решения. Остальные три фактора имеют более низкое влияние на принятие решения.
