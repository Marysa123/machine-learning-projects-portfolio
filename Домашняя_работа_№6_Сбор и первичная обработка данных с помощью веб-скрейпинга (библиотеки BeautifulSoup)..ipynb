## Задача 1. Книги без буквы E.

**Условие.** В истории литературы известен случай написания романа объемом около
50 тыс. слов, в котором ни разу не была употреблена самая популярная
в английском алфавите буква E. Название его – «Gadsby».
Напишите программу, которая будет считывать список слов из файла и собирать статистику о том, в каком проценте слов используется каждая буква алфавита. Выведите результат для всех 26 букв английского алфавита и отдельно отметьте букву, которая встречалась в словах наиболее редко. В  вашей программе должны игнорироваться знаки препинания и регистр символов.


**Подсказки.**
1. Переведите все слова в верхний регистр:
```
word = word.upper().rstrip() # upper - переводит строку в верхний регистр
```
2. Создайте словарь со счетчиком слов, содержащий каждую букву
```
# Для каждой буквы инициализируем счетчик нулем
counts = {}
for ch in "ABCDEFGHIJKLMNOPQRSTUVWXYZ":
       counts[ch] = 0
```
3. Наиболее редко используемые буквы можно найти следующим образом:
```
smallest_count = min(counts.values()) # min - возвращает наименьшее число в списке
```

````python
from decimal import Decimal;

file = open("input.txt", "w") # мы открыли файл для записи, если он уже создан, то будет перезаписан

text = 'English texts for beginners to practice reading and comprehension online and for free. Practicing jyourq comprehension of written English will both improve your vocabulary and understanding of grammar and word order. The texts below are designed to help you develop while giving you an instant evaluation of your progress. Prepared by experienced English teachers, the texts, articles and conversations are brief and appropriate to your level of proficiency. Take the multiple-choice quiz following each text, and you ll get the results immediately. You will feel both challenged and accomplished! You can even download (as PDF) and print the texts and exercises. It s enjoyable, fun and free. Good luck!'
# мы получили в переменную file дескриптор файла (ссылку на файл) и теперь по нему можем работать с файлом
file.write(text) # Записываем текст целиком в файл
file.close() # Закрываем файл

file = open("input.txt", "r") # Теперь мы открываем файл для чтения

lines = file.read() # Извлекаем целиком текст из файла

# Убираем все небуквенные символы из словов
word = ''.join(filter(str.isalpha, lines.lower()))

word = word.upper() # upper - переводит строку в верхний регистр

total_letters = len(word)
print(total_letters)

counts = {'A': 0,'B': 0,'C': 0,'D': 0,'E': 0,'F': 0,'G': 0,'H': 0,'I': 0,'J': 0,'K': 0,'L': 0,'M': 0,'N': 0,'O': 0,'P': 0,'Q': 0,'R': 0,'S': 0,'T': 0,'U': 0,'V': 0,'W': 0,'X': 0,'Y': 0,'Z': 0}
for ch in word:
    counts[ch] += 1

print(counts)

for ch in 'ABCDEFGHIJKLMNOPQRSTUVWXYZ':
    counts[ch] = round(((counts.get(ch) / total_letters) * 100), 2)


for key, value in counts.items():
  print("{0}: {1}".format(key,value),'%')


smallest_count = min(counts.values()) # min - возвращает наименьшее число в списке

for key in counts:
    if counts[key] == smallest_count:
        print("Буква, которая встречается крайне мало:", key, smallest_count,'%')
        break
````
