## Задача 2. Сувениры и безделушки.

после чего выведите на экран общий вес посылки.


```python
#Создадим переменные, которые будут принимать значения с клавиатуры
souvenirs_quantity = int(input("Введите количество, купленных вами сувениров: "))
trinkets_quantity = int(input("Введите количество, купленных вами безделушек: "))

#Создадим переменные, которые будут хранить постоянные значения веса данных предметов
souvenirs_weight = 75
trinkets_weight = 112

#Вычисляем общую сумму веса каждого наименования
souvenirs_total_weight = souvenirs_quantity * souvenirs_weight
trinkets_total_weight = trinkets_quantity * trinkets_weight

#Вычисляем общую сумму веса всей посылки
total_sum_weight = trinkets_total_weight + souvenirs_total_weight

formatted_amount = "{:.2f} грамм".format(total_sum_weight)

#Выводим на экран
print("Общий вес посылки составляет: ",formatted_amount)


#Сократим код
souvenirs_total_weight = int(input("Введите количество, купленных вами сувениров: ")) * 75
trinkets_total_weight = int(input("Введите количество, купленных вами безделушек: ")) * 112

total_sum_Weight = trinkets_total_weight + souvenirs_total_weight

print("Общий вес посылки составляет: ",total_sum_Weight)


#Сократим код
total_sum_weight = int(input("Введите количество, купленных вами безделушек: ")) * 112 + int(input("Введите количество, купленных вами сувениров: ")) * 75
print("Общий вес посылки составляет: ",total_sum_weight,"грамм")


#Сократим код
print("Общий вес посылки составляет: ",int(input("Введите количество, купленных вами безделушек: ")) * 112 + int(input("Введите количество, купленных вами сувениров: ")) * 75,"грамм")
```

## Задача 3. Сложные проценты

**Условие.** Представьте, что вы открыли в банке сберегательный счет под 4 % годовых. Проценты банк рассчитывает в конце года и добавляет к сумме счета.
Напишите программу, которая запрашивает у пользователя сумму первоначального депозита, после чего рассчитывает и выводит на экран сумму
на счету в конце первого, второго и третьего годов. Все суммы должны
быть округлены до двух знаков после запятой.



```python
#Создадим переменную, которая будут принимать значение с клавиатуры
initial_deposit = float(input("Здравствуйте, введите сумму вашего первоначального депозита: "))

#Высчитывает сумму на конец первого года
initial_deposit += 0.04 * initial_deposit
print("Сумма на счёте в конце первого года:","{:.2f} рублей".format(initial_deposit))

#Высчитывает сумму на конец второго года
initial_deposit += 0.04 * initial_deposit
print("Сумма на счёте в конце второго года:","{:.2f} рублей".format(initial_deposit))

#Высчитывает сумму на конец третьего года
initial_deposit += 0.04 * initial_deposit
print("Сумма на счёте в конце третьего года:","{:.2f} рублей".format(initial_deposit))

print("Общая сумма за все три года:","{:.2f} рублей".format(initial_deposit))

```

## Задача 4. Арифметика

**Условие.** Создайте программу, которая запрашивает у пользователя два целых числа a и b, после чего выводит на экран результаты следующих математических операций:
* сумма a и b;
* разница между a и b;
* произведение a и b;
* частное от деления a на b;
* остаток от деления a на b;
* десятичный логарифм числа a;
* результат возведения числа a в степень b.

```python
import math
#Создадим переменные, которые будут принимать значения с клавиатуры
a = int(input("Введите значение переменной a: "))
b = int(input("Введите значение переменной b: "))

print("Сумма a и b =", a+b)
print("Разница между a и b =", a - b)
print("Произведение a и b = ", a*b)
print("Частное от деления a на b =", a/b)
print("Остаток от деления a на b =", a%b)
print("Десятичный логарифм числа a = ", math.log(a))
print("Результат возведения числа a в степень b =", a**b)
```

## Задача 5. Размен

**Условие.** Представьте, что вы пишете программное обеспечение для автоматической кассы в  магазине самообслуживания. Одной из функций, заложенных в кассу, должен быть расчет сдачи в случае оплаты покупателем наличными.
Напишите программу, которая будет запрашивать у пользователя сумму сдачи в копейках. После этого она должна рассчитать и вывести на экран, сколько и каких монет потребуется для выдачи указанной суммы, при условии что должно быть задействовано минимально возможное количество монет. Допустим, у нас есть в распоряжении монеты достоинством в 1, 5, 10, 50 копеек, а также в 1, 2 и 5 рублей.


```python
#Создадим переменную, которая будут принимать значение с клавиатуры
received_coins = float(input("Полученные деньги: "))
total_purchase_amount = float(input("Общая сумма покупки составила: "))

remains = total_purchase_amount - received_coins
print("Ваша сдача: ","{:.2f} рублей".format(-remains))

#Деление общей суммы сдачи на 5 для получения количества монет в номинале 5 рублей, которые необходимо выдать
delivery_of_5 = -int(remains / 5)

#Получение остатка от деления на 5 для дальнейшего подсчета выдачи монет
delivery_of_5_och = -int(remains) % 5

#Деление остатка от деления для получения количества монет в номинале 2 рубля
delivery_of_2 = -int(delivery_of_5_och // 2)

#Получение остатка от деления на 2 для дальнейшего подсчета выдачи монет
delivery_of_2_och = -int(delivery_of_5_och % 2)

#Получение остатка от деления на 2 для получения количества монет в номинале 1 рубль
delivery_of_1 = -int(delivery_of_2_och // 1)


print("Необходимо выдать сдачу 5 рублёвыми монетами в количестве: ",delivery_of_5)
print("Необходимо выдать сдачу 2 рублёвыми монетами в количестве: ",-delivery_of_2)
print("Необходимо выдать сдачу 1 рублёвыми монетами в количестве: ",delivery_of_1)


#Подсчёт количества копеек
fractional_part = -remains - int(-remains)
print("Количество копеек для выдачи сдачи: ","{:.2f} копеек".format(fractional_part))

#Деление общей суммы копеек со сдачи на 50 для получения количества копеек в номинале 50
delivery_of_50 = int(fractional_part / 0.5)

#Округление значений полсе запятой
fractional_part = round(fractional_part,2)

#Получение остатка от деления на 50 для дальнейшего подсчета выдачи монет
delivery_of_50_och = float(fractional_part) % 0.5

#Округление значений полсе запятой
delivery_of_50_och = round(delivery_of_50_och,2)

#Деление остатка от деления для получения количества монет в номинале 10 копеек
delivery_of_10 = int(delivery_of_50_och // 0.1)

#Получение остатка от деления для получения количества монет в номинале 10 копеек
delivery_of_05_och = delivery_of_50_och % 0.1

#Округление значений полсе запятой
fractional_part = round(delivery_of_05_och,2)

#Получение остатка от деления для получения количества монет в номинале 5 копеек
delivery_of_05 = int(fractional_part // 0.05)

#Получение остатка от деления для получения количества монет в номинале 1 копейки
delivery_of_01_och = fractional_part % 0.05

#Округление значений полсе запятой
fractional_part = round(delivery_of_01_och,2)

#Получение остатка от деления для получения количества монет в номинале 1 копейки
delivery_of_01 = int(fractional_part / 0.01)


print("Необходимо выдать сдачу 50 копейками в количестве: ",delivery_of_50,"монет")
print("Необходимо выдать сдачу 10 копейками в количестве: ",delivery_of_10,"монет")
print("Необходимо выдать сдачу 5 копейками в количестве: ",delivery_of_05,"монет")
print("Необходимо выдать сдачу 1 копейками в количестве: ",delivery_of_01,"монет")









```
